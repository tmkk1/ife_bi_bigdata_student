## 🏭 2. Pourquoi Python pour les Ingénieurs de Production ? (Échelle & Reproductibilité)
Dans les ateliers traditionnels, les équipes s'appuient couramment sur Microsoft Excel pour comptabiliser les rebuts et suivre l'utilisation machine. Bien qu'Excel rende service pour des vérifications rapides, il s'effondre face aux exigences de l'Industrie 4.0 :

1. **Plafond Strict du Nombre de Lignes :** Excel est incapable d'ouvrir un fichier contenant plus de **1 048 576 lignes**. Une presse d'emboutissage moderne échantillonnant sa télémesure toutes les 10 millisecondes atteint cette limite en moins de 3 heures !
2. **Altération Silencieuse des Formules :** Lorsqu'une formule est accidentellement écrasée dans la cellule `F42`, l'erreur demeure indétectable jusqu'à ce qu'un tableau de bord exécutif publie des indicateurs faux. En Python, chaque transformation est explicite, reproductible et sous contrôle de version.
3. **Connecteur Universel Multi-Sources :** Python agrège aisément les flux CSV de supervision SCADA, les tickets JSON de GMAO, les bases relationnelles SQL d'ERP et les classeurs Excel au sein d'une chaîne automatisée unique.

Testons la vitesse et l'efficacité mémoire de Python en générant **1 500 000 mesures de capteurs synthétiques** directement en mémoire vive :


````python
import time
import sys

# Génération de 1,5 million de mesures simulées de vibration (mm/s)
start_time = time.time()
vibrations_simulees = [round(2.5 + (i % 100) * 0.05, 3) for i in range(1_500_000)]
elapsed_time = time.time() - start_time

memoire_mo = sys.getsizeof(vibrations_simulees) / (1024 * 1024)

print(f"✅ {len(vibrations_simulees):,} mesures de capteurs générées en {elapsed_time:.4f} secondes !")
print(f"📊 Mémoire allouée : {memoire_mo:.2f} Mo")
print(f"🔍 Aperçu des 5 premières mesures : {vibrations_simulees[:5]}")
print(f"⚡ Limite maximale d'Excel (1 048 576) dépassée de : {len(vibrations_simulees) - 1_048_576:,} lignes !")
````

## 🔢 4. Types de Données Primitifs & Transtypage Industriel
Python s'appuie sur cinq types primitifs incontournables présents dans tous les flux de supervision SCADA, ERP et capteurs IoT :

| Type | Désignation | Contexte Industriel Typique | Exemples |
| :--- | :--- | :--- | :--- |
| `int` | Entier (*Integer*) | Comptages discrets de pièces, cycles machine, numéros de poste | `42`, `-1`, `1500` |
| `float` | Flottant (*Floating-point*) | Mesures physiques continues : couple (Nm), temp. (°C), puissance (kW) | `318.5`, `-999.0`, `14.25` |
| `str` | Chaîne de texte (*String*) | Identifiants d'équipement, codes d'erreur, horodatages bruts | `"PRESS_01"`, `"ERR_HYD_404"` |
| `bool` | Booléen (*Boolean*) | Indicateurs binaires d'état, verrouillages, conformité qualité | `True`, `False` |
| `NoneType`| Nul (*None / Null*) | Perte de signal capteur, champ non renseigné, arrêt réseau | `None` |

### ⚠️ Le Piège de la « Télémesure en Chaîne de Texte »
De nombreux automates programmables industriels (API / PLC) et passerelles IoT transmettent leurs mesures sur liaison série ou protocole MQTT sous forme de chaînes de caractères ASCII (ex. `"142.75"` au lieu de la grandeur numérique `142.75`). Si vous tentez une opération arithmétique sur une chaîne de texte, Python déclenchera immédiatement une erreur `TypeError` !

Exécutez la cellule ci-dessous pour observer l'inspection de types et le transtypage (*type casting*) :


````python
# Inspection des types avec type()
lecture_capteur_brute = "142.75"  # Transmis sous forme de chaîne de caractères sur le réseau
pieces_produites_str = "850"

print("Type initial de lecture_capteur_brute :", type(lecture_capteur_brute))
print("Type initial de pieces_produites_str : ", type(pieces_produites_str))

# Transtypage (Conversion de type) :
capteur_float = float(lecture_capteur_brute)
pieces_int = int(pieces_produites_str)

print(f"\nFlottant converti : {capteur_float} (Type : {type(capteur_float).__name__})")
print(f"Entier converti   : {pieces_int} (Type : {type(pieces_int).__name__})")

# Les opérations mathématiques fonctionnent désormais correctement :
energie_totale_kwh = capteur_float * 1.5
print(f"Énergie calculée : {energie_totale_kwh:.2f} kWh")
````

## 🗃️ 5. Collections Fondamentales : Listes & Dictionnaires
En ingénierie des données, les grandeurs scalaires existent rarement isolément ; elles sont agrégées au sein de collections. Les deux structures piliers sont les **Listes** et les **Dictionnaires**.

### 1. Les Listes : Séquences Ordonnées et Modifiables
- Les listes sont délimitées par des crochets `[ ... ]`.
- Les éléments sont indexés à partir de `0` (indexation base zéro).
- Elles supportent l'indexation négative (`-1` désigne le dernier élément), le découpage par tranches (*slicing*, ex. `[0:2]`) et l'ajout dynamique (`.append()`) :


````python
# Manipulation de listes dans un atelier d'emboutissage
presses_emboutissage = ["PRESS_01", "PRESS_02", "PRESS_03"]

# Ajout d'une nouvelle machine nouvellement mise en service :
presses_emboutissage.append("PRESS_04")

print("Parc complet de machines :    ", presses_emboutissage)
print("Première machine (index [0]) :", presses_emboutissage[0])
print("Dernière machine (index [-1]):", presses_emboutissage[-1])
print("Tranche de sous-ligne [1:3] : ", presses_emboutissage[1:3])
print("Nombre total de machines :    ", len(presses_emboutissage))
````

### 2. Les Dictionnaires : Associations Clés-Valeurs (Le Modèle Mental du JSON)
- Les dictionnaires sont délimités par des accolades `{ "clé": valeur }`.
- Ils assurent un accès rapide par nom de clé plutôt que par position numérique.
- **Concept Majeur :** En Python, un dictionnaire reproduit rigoureusement la structure d'un **enregistrement JSON** (*JavaScript Object Notation*), format universel des APIs REST, des webhooks et de la télémesure SCADA moderne !


````python
# Dictionnaire de télémesure pour une machine individuelle
telemesure_machine = {
    "machine_id": "PRESS_01",
    "nominal_cycle_sec": 12.0,
    "operating_cost_eur": 180.0,
    "is_operational": True,
    "oil_temperature_c": 74.2
}

# Accès aux valeurs par clé :
print(f"Identifiant Machine : {telemesure_machine['machine_id']}")
print(f"Coût Opérationnel :   {telemesure_machine['operating_cost_eur']} EUR/h")

# Accès défensif via .get() : Renvoie une valeur par défaut si la clé est absente (évite les plantages KeyError !)
niveau_vibration = telemesure_machine.get("vibration_rms", 0.0)
print(f"Niveau de Vibration : {niveau_vibration} mm/s (Valeur de secours si capteur absent)")

# Modification et enrichissement de champs :
telemesure_machine["is_operational"] = False
telemesure_machine["last_maintenance_date"] = "2026-10-15"
print("Fiche actualisée :    ", telemesure_machine)
````

## ⚙️ 6. Fonctions : Logique d'Ingénierie Réutilisable
En gestion de production, des calculs tels que le taux de rebut, le rendement de cycle et le Taux de Rendement Synthétique (TRS / *OEE*) doivent être exécutés de façon répétée sur des milliers de postes et de lots.  
Coder ces formules en dur dans des scripts dispersés conduit inévitablement à des **dérives de calcul** (deux services calculant l'indicateur avec des règles contradictoires).

Les fonctions isolent rigoureusement la logique métier :
- Déclarées avec le mot-clé `def`.
- Définissent des paramètres dotés d'**Indications de Type (*Type Hints*)** (`param: type -> type_retour`) formalisant les entrées attendues.

$$TRS = \text{Disponibilité} \times \text{Performance} \times \text{Qualité}$$

1. **Facteur de Disponibilité ($A$) :**
$$A = \frac{\text{Temps de Fonctionnement Réel (min)}}{\text{Temps Prévu de Production (min)}}$$

$$P = \frac{\text{Temps de Cycle Idéal (sec)} \times \text{Nombre Total de Pièces Produites}}{\text{Temps de Fonctionnement Réel (min)} \times 60\,\text{sec/min}}$$
3. **Facteur de Qualité ($Q$) :**
$$Q = \frac{\text{Pièces Totales Produites} - \text{Pièces Rebutées}}{\text{Pièces Totales Produites}} = \frac{\text{Pièces Conformes}}{\text{Pièces Totales Produites}}$$
