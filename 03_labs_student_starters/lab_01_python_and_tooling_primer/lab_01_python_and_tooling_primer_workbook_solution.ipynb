### 🛠️ Challenge 1: Hydraulic Press Safety Interlock (Solution)


```python
def evaluate_press_safety(hydraulic_pressure_bar: float, safety_curtain_active: bool) -> str:
    """Evaluates hydraulic press safety interlocks based on pressure and optical curtain."""
    if not safety_curtain_active:
        return "E-STOP: Safety curtain breached! Machine halted immediately."
    elif hydraulic_pressure_bar > 250.0:
        return "ALARM: Overpressure detected (>250 bar)! Relief valve opened."
    elif hydraulic_pressure_bar < 160.0:
        return "WARNING: Low pressure (<160 bar). Check pump and oil reservoir."
    else:
        return "NOMINAL: Pressure within normal operating envelope (160-250 bar)."

# Automated Verification Test Suite
test_1 = evaluate_press_safety(hydraulic_pressure_bar=200.0, safety_curtain_active=False)
test_2 = evaluate_press_safety(hydraulic_pressure_bar=265.0, safety_curtain_active=True)
test_3 = evaluate_press_safety(hydraulic_pressure_bar=140.0, safety_curtain_active=True)
test_4 = evaluate_press_safety(hydraulic_pressure_bar=210.0, safety_curtain_active=True)

print("Test 1 (Breached curtain):", test_1)
print("Test 2 (Overpressure):    ", test_2)
print("Test 3 (Low pressure):     ", test_3)
print("Test 4 (Nominal):          ", test_4)

assert "E-STOP" in test_1, "Test 1 failed!"
assert "ALARM" in test_2, "Test 2 failed!"
assert "WARNING" in test_3, "Test 3 failed!"
assert "NOMINAL" in test_4, "Test 4 failed!"
print("\n🎉 Challenge 1 Passed! All safety interlock conditions verified.")
```

### 🛠️ Challenge 2: Parsing an IoT Telemetry Packet String (Solution)


```python
raw_packet = "PRESS_03,1250,94.35,True,None"

# 1. Split packet into list of string tokens
tokens = raw_packet.split(",")

# 2. Parse and cast tokens
machine_id = str(tokens[0])
cycles_completed = int(tokens[1])
motor_power_kw = float(tokens[2])
guard_interlock_ok = tokens[3].strip() == "True"
active_error_code = None if tokens[4].strip() == "None" else tokens[4].strip()

# 3. Calculate energy consumed in 45 minutes (0.75 hours)
energy_consumed_kwh = round(motor_power_kw * (45.0 / 60.0), 4)

print(f"Machine ID:           {machine_id} (type: {type(machine_id).__name__})")
print(f"Cycles Completed:     {cycles_completed} (type: {type(cycles_completed).__name__})")
print(f"Motor Power (kW):     {motor_power_kw} (type: {type(motor_power_kw).__name__})")
print(f"Guard Interlock OK:   {guard_interlock_ok} (type: {type(guard_interlock_ok).__name__})")
print(f"Active Error Code:    {active_error_code} (type: {type(active_error_code).__name__})")
print(f"Energy Consumed:      {energy_consumed_kwh} kWh")

# Verification assertions
assert machine_id == "PRESS_03"
assert cycles_completed == 1250
assert abs(motor_power_kw - 94.35) < 1e-4
assert guard_interlock_ok is True
assert active_error_code is None
assert abs(energy_consumed_kwh - 70.7625) < 1e-3
print("\n🎉 Challenge 2 Passed! Telemetry string correctly parsed and typed.")
```

```python
# A fleet represented as a list of machine records
shopfloor_fleet = [
    {"machine_id": "PRESS_01", "type": "Mechanical Press", "operating_hours": 1250, "oil_temp_c": 72.0, "status": "RUNNING"},
    {"machine_id": "PRESS_02", "type": "Hydraulic Press",  "operating_hours": 2400, "oil_temp_c": 93.5, "status": "RUNNING"},
    {"machine_id": "PRESS_03", "type": "Servo Press",      "operating_hours": 890,  "oil_temp_c": 58.0, "status": "IDLE"},
    {"machine_id": "PRESS_04", "type": "Hydraulic Press",  "operating_hours": 3150, "oil_temp_c": 96.2, "status": "RUNNING"},
    {"machine_id": "PRESS_05", "type": "Mechanical Press", "operating_hours": 1500, "oil_temp_c": 64.0, "status": "MAINTENANCE"}
]

print(f"Loaded {len(shopfloor_fleet)} machine telemetry records.")
```

### 🛠️ Challenge 3: Fleet Maintenance Triage (Solution)


```python
# Solution for Challenge 3
triage_machine_ids = []
running_hours_list = []

for machine in shopfloor_fleet:
    if machine["status"] == "RUNNING":
        running_hours_list.append(machine["operating_hours"])
        if machine["oil_temp_c"] > 90.0 or machine["operating_hours"] > 2000:
            triage_machine_ids.append(machine["machine_id"])

avg_running_hours = round(sum(running_hours_list) / len(running_hours_list), 2)

print(f"Machines requiring immediate triage: {triage_machine_ids}")
print(f"Average operating hours (RUNNING):   {avg_running_hours} hours")

# Assertions
assert triage_machine_ids == ["PRESS_02", "PRESS_04"], "Triage machine list incorrect!"
assert avg_running_hours == 2266.67, f"Expected 2266.67, got {avg_running_hours}"
print("\n🎉 Challenge 3 Passed! Shopfloor triage logic fully verified.")
```

## ⚙️ 6. Functions: Reusable Engineering Logic


```python
# Example from Primer: Defensive calculation of scrap percentage
def calculate_scrap_rate(produced_units: int, scrap_units: int) -> float:
    """Calculates scrap percentage with defensive division-by-zero guard."""
    if produced_units <= 0:
        return 0.0
    return round((scrap_units / produced_units) * 100.0, 2)

# Test with standard shift output:
shift_1_scrap = calculate_scrap_rate(produced_units=1500, scrap_units=45)
shift_zero_scrap = calculate_scrap_rate(produced_units=0, scrap_units=0)

print(f"Shift 1 Scrap Rate:    {shift_1_scrap}%")
print(f"Zero Production Scrap: {shift_zero_scrap}% (No ZeroDivisionError!)")
```

### 🛠️ Challenge 4: Industrial Overall Equipment Effectiveness (OEE) Engine (Solution)


```python
def calculate_oee(
    planned_time_min: float,
    actual_operating_time_min: float,
    ideal_cycle_sec: float,
    total_parts: int,
    scrap_parts: int
) -> dict:
    """Calculates OEE and component factors with defensive guards."""
    if planned_time_min <= 0 or actual_operating_time_min <= 0 or total_parts <= 0:
        return {
            "availability": 0.0,
            "performance": 0.0,
            "quality": 0.0,
            "oee": 0.0,
            "oee_pct": 0.0,
            "classification": "UNACCEPTABLE"
        }
    
    # 1. Availability
    availability = min(1.0, actual_operating_time_min / planned_time_min)
    
    # 2. Performance
    operating_time_sec = actual_operating_time_min * 60.0
    ideal_production_time_sec = total_parts * ideal_cycle_sec
    performance = min(1.0, ideal_production_time_sec / operating_time_sec)
    
    # 3. Quality
    good_parts = max(0, total_parts - scrap_parts)
    quality = min(1.0, good_parts / total_parts)
    
    # 4. Composite OEE
    oee = availability * performance * quality
    oee_pct = oee * 100.0
    
    # 5. Classification
    if oee_pct >= 85.0:
        classification = "WORLD_CLASS"
    elif oee_pct >= 70.0:
        classification = "ACCEPTABLE"
    else:
        classification = "UNACCEPTABLE"
        
    return {
        "availability": round(availability, 4),
        "performance": round(performance, 4),
        "quality": round(quality, 4),
        "oee": round(oee, 4),
        "oee_pct": round(oee_pct, 2),
        "classification": classification
    }

# Test shift calculation:
# 8-hour shift = 480 min planned, 420 min operating, 12-sec ideal cycle, 1800 parts, 45 scrap
shift_kpi = calculate_oee(
    planned_time_min=480.0,
    actual_operating_time_min=420.0,
    ideal_cycle_sec=12.0,
    total_parts=1800,
    scrap_parts=45
)

print("Shift OEE Report:")
for k, v in shift_kpi.items():
    print(f"  {k:15}: {v}")

# Verification Assertions
assert shift_kpi["availability"] == 0.875
assert shift_kpi["performance"] == 0.8571
assert shift_kpi["quality"] == 0.975
assert shift_kpi["oee"] == 0.7312
assert shift_kpi["oee_pct"] == 73.12
assert shift_kpi["classification"] == "ACCEPTABLE"
print("\n🎉 Challenge 4 Passed! Industrial OEE engine verified.")
```
