## 🏭 2. Why Python for Production Engineers? (Scale & Reproducibility)
In traditional factory offices, engineers often rely on Microsoft Excel for scrap tracking and machine utilization. While Excel is helpful for quick checks, it fails catastrophically when an enterprise scales up:

1. **Row Limits:** Excel cannot open any file with more than **1,048,576 rows**. A modern stamping press recording sensor telemetry every 10 milliseconds hits that limit in less than 3 hours!
2. **Hidden Formula Corruption:** When a formula is accidentally overwritten in cell `F42`, the error remains silent until an executive dashboard produces false metrics. In Python, every transformation is explicit, testable, and version-controlled.
3. **Multi-Source Glue:** Python effortlessly ingests SCADA CSV logs, CMMS JSON tickets, ERP SQL databases, and Excel master files into a single automated pipeline.

## 🔢 4. Primitive Data Types & Industrial Type Casting
Python has five fundamental primitive data types that you will encounter across every SCADA, ERP, and IoT data feed:

| Type | Name | Typical Industrial Context | Examples |
| :--- | :--- | :--- | :--- |
| `int` | Integer | Discrete item counts, cycle numbers, shift IDs | `42`, `-1`, `1500` |
| `float` | Floating-point | Continuous physical measurements: torque (Nm), temp (°C), power (kW) | `318.5`, `-999.0`, `14.25` |
| `str` | String (Text) | Equipment identifiers, failure mode codes, unparsed timestamps | `"PRESS_01"`, `"ERR_HYD_404"` |
| `bool` | Boolean | Binary operational flags, interlocks, quality checks | `True`, `False` |
| `NoneType`| None (Null) | Missing sensor signals, transmission dropouts, unrecorded telemetry | `None` |

### ⚠️ The "String Telemetry Trap"
Many industrial PLCs and IoT gateways transmit measurements over serial or MQTT as ASCII strings (e.g., `"142.75"` instead of the number `142.75`). If you attempt to do arithmetic on text strings, Python will raise a `TypeError`!

Run the cell below to inspect types and practice type conversion (casting):


````python
# Inspecting types with type()
raw_sensor_reading = "142.75"  # Sent as string over serial
parts_produced_str = "850"

print("Type of raw_sensor_reading:", type(raw_sensor_reading))
print("Type of parts_produced_str:", type(parts_produced_str))

# Type Casting (Conversion):
sensor_float = float(raw_sensor_reading)
parts_int = int(parts_produced_str)

print(f"\nCasted float: {sensor_float} (Type: {type(sensor_float)})")
print(f"Casted int:   {parts_int} (Type: {type(parts_int)})")

# Now mathematical calculations work properly:
total_energy_kwh = sensor_float * 1.5
print(f"Calculated Energy: {total_energy_kwh:.2f} kWh")
````

## 🗃️ 5. Core Collections: Lists & Dictionaries
In production data engineering, physical measurements rarely exist in isolation; they are grouped into collections. The two foundational structures are **Lists** and **Dictionaries**.

- Lists are enclosed in square brackets `[ ... ]`.

- Support negative indexing (`-1` is the last item), slicing (`[0:2]`), and appending (`.append()`):


````python
# Working with Lists in an automotive stamping shop
stamping_presses = ["PRESS_01", "PRESS_02", "PRESS_03"]

# Append a newly commissioned machine:
stamping_presses.append("PRESS_04")

print("All machines:       ", stamping_presses)
print("First machine ([0]):", stamping_presses[0])
print("Last machine ([-1]):", stamping_presses[-1])
print("Sub-line slice [1:3]:", stamping_presses[1:3])
print("Fleet size count:    ", len(stamping_presses))
````

### 2. Dictionaries: Key-Value Mappings (The Mental Model for JSON)
- Dictionaries are enclosed in curly braces `{ "key": value }`.
- Provide fast lookup by named keys rather than numeric indices.
- **Critical Concept:** Dictionaries in Python have the exact same structure as **JSON records** (JavaScript Object Notation), the standard format used by REST APIs, webhooks, and modern SCADA telemetry!


````python
# Telemetry dictionary for a single machine
machine_telemetry = {
    "machine_id": "PRESS_01",
    "nominal_cycle_sec": 12.0,
    "operating_cost_eur": 180.0,
    "is_operational": True,
    "oil_temperature_c": 74.2
}

# Accessing values by key:
print(f"Machine ID:       {machine_telemetry['machine_id']}")
print(f"Operating Cost:   {machine_telemetry['operating_cost_eur']} EUR/h")

# Defensive access with .get(): Returns default value if key is missing (avoids KeyError crashes!)
vibration_reading = machine_telemetry.get("vibration_rms", 0.0)
print(f"Vibration Level:  {vibration_reading} mm/s (Default fallback used if sensor missing)")

# Modifying and adding fields:
machine_telemetry["is_operational"] = False
machine_telemetry["last_maintenance_date"] = "2026-10-15"
print("Updated record:   ", machine_telemetry)
````

## ⚙️ 6. Functions: Reusable Engineering Logic
In manufacturing operations, calculations like scrap rate, cycle time efficiency, and Overall Equipment Effectiveness (OEE) must be executed repeatedly across thousands of workcenters and batches.  
Hardcoding these formulas in different scripts or spreadsheet tabs leads to **formula drift** (where two teams calculate OEE differently).

Functions encapsulate engineering logic cleanly:
- Defined with the `def` keyword.
- Accept parameters with optional **Type Hints** (`param: type -> return_type`) that document expected inputs.
- Include a **Docstring** (triple quotes) explaining what the function calculates.
- Return explicit results via `return`.


````python
# Example from Primer: Defensive calculation of scrap percentage
def calculate_scrap_rate(produced_units: int, scrap_units: int) -> float:
    """Calculates scrap percentage with defensive division-by-zero guard."""
    if produced_units <= 0:
        return 0.0
    return round((scrap_units / produced_units) * 100.0, 2)

# Test with standard shift output:
shift_1_scrap = calculate_scrap_rate(produced_units=1500, scrap_units=45)
shift_zero_scrap = calculate_scrap_rate(produced_units=0, scrap_units=0)

print(f"Shift 1 Scrap Rate:    {shift_1_scrap}%")
print(f"Zero Production Scrap: {shift_zero_scrap}% (No ZeroDivisionError!)")
````

1. **Availability Factor ($A$):**
$$A = \frac{\text{Actual Operating Time (min)}}{\text{Planned Production Time (min)}}$$
2. **Performance Factor ($P$):**
$$P = \frac{\text{Ideal Cycle Time (sec)} \times \text{Total Parts Produced}}{\text{Actual Operating Time (min)} \times 60\,\text{sec/min}}$$
3. **Quality Factor ($Q$):**
$$Q = \frac{\text{Total Parts Produced} - \text{Scrap Parts}}{\text{Total Parts Produced}} = \frac{\text{Good Parts}}{\text{Total Parts Produced}}$$
